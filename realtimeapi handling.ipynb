```python
import pandas as pd
import requests
```

```python
# Fetching earthquake data from the USGS API
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_day.geojson"
response = requests.get(url)
data = response.json()
```

```python
#Display the keys of API through which we can access
print(data.keys())
```

```python
#To view the single value of earthquake
single_earthquake = data["features"][0]
single_earthquake
```

```python
#To access Multiple values
earthquakes = data["features"]
view = pd.json_normalize(earthquakes)
view


```

```python
# Viewing the cleaned data According to the features 
clean_view=view[['properties.place' , 'properties.time']]
clean_view
```

```python
# Viewing the first few rows of the cleaned data    
clean_view.head()
```

```python
# Viewing the last few rows of the cleaned data
clean_view.tail()
```
